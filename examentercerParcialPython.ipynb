- Escribir un generador que genere todos los factores de un número. Un factor es un número natural que puede dividir exactamente a otro número.

```python
# Your solution HERE !!!!

def generador_factores(n):
  
# Validamos que sea un número natural
    if n <= 0:
        print("Por favor, ingresa un número natural mayor a 0.")
        return

    for i in range(1, n + 1):
        if n % i == 0:
            yield i

numero = 18
print(f"Los factores de {numero} son:")

# Iteramos sobre el generador, es decir se repite hasta que concluye
for factor in generador_factores(numero):
    print(factor)
```

- Escribe una función que reciba un DataFrame, le agregue una nueva columna llamada ‘Promedio’ que sea el promedio de sus columnas Luego, filtre las filas donde el valor de ‘Promedio’ sea mayor que 5 y lo ordene por la columna ‘Promedio’ en orden descendente.

```python
# Your solution HERE !!!!

import pandas as pd

def procesar_datos(df):
    
# 1. Crear la columna 'Promedio' calculando la media de las filas (axis=1)
    # numeric_only=True asegura que solo se promedien columnas con números
    df['Promedio'] = df.mean(axis=1, numeric_only=True)
    
# 2. Filtrar filas donde el valor de 'Promedio' sea mayor que 5
    df_filtrado = df[df['Promedio'] > 5]
    
# 3. Ordenar por 'Promedio' en orden descendente
    df_final = df_filtrado.sort_values(by='Promedio', ascending=False)
    
    return df_final

# Conjunto de datos de prueba (Notas de alumnos)
datos = {
    'Inglés': [8, 4, 9, 2, 7, 10],
    'Estadística': [7, 5, 8, 3, 6, 9],
    'Programación': [9, 3, 10, 4, 5, 10]
}

df_inicial = pd.DataFrame(datos)

# Aplicar función
df_resultado = procesar_datos(df_inicial)

# Resultado
print("DataFrame Final Procesado:")
print(df_resultado)
```

## 03 - 1.25pt

- Dado el siguiente DataFrame, reemplaza los valores nulos en la columna ‘B’ con la media de los valores de esa columna. Luego, normaliza los valores de la columna ‘C’ para que estén en el rango [0, 1].

```python
# Your solution HERE !!!!

import pandas as pd

# Datos de entrada
source_dataframe = pd.DataFrame({
    'A': [1, 2, 3], 
    'B': [4, None, 6], 
    'C': [7, 8, 9]
})

# 1. Reemplazar nulos en 'B' con la media de 'B', la media de 4 y 6 es 5.
source_dataframe['B'] = source_dataframe['B'].fillna(source_dataframe['B'].mean())

# 2. Normalizar columna 'C' al rango [0, 1]
# Aplicamos la fórmula: (valor - min) / (max - min)
min_c = source_dataframe['C'].min()
max_c = source_dataframe['C'].max()

source_dataframe['C'] = (source_dataframe['C'] - min_c) / (max_c - min_c)

# Resultado final
print(source_dataframe)
```

```python
# Your solution HERE !!!!
# 05 - Representar una figura con tres gráficas con tres diferentes tasas el total nacional y ambos sexos, 
#de manera que en cada subfigura se represente la evolución por trimestre de cada año (cada año se considera una linea con un color diferente).



import matplotlib.pyplot as plt

# 1. Columnas de Año y Trimestre
df_limpio['Año'] = df_limpio['Periodo'].str[:4]
df_limpio['Trimestre'] = df_limpio['Periodo'].str[-2:]

# 2. Filtros por Total Nacional y Ambos sexos
df_nacional = df_limpio[(df_limpio['Provincias'] == 'Total Nacional') & 
                        (df_limpio['Sexo'] == 'Ambos sexos')]

# 3. Definir las 3 tasas y los años que queremos pintar
lista_tasas = ['Tasa de actividad', 'Tasa de paro de la población', 'Tasa de empleo de la población']
lista_anios = sorted(df_nacional['Año'].unique())

# 4. Crear la figura por cada tasa
plt.figure(figsize=(20, 6))

for i in range(len(lista_tasas)):
    # Elegimos el cuadro actual (1, 2 o 3)
    plt.subplot(1, 3, i + 1)
    
    tasa_actual = lista_tasas[i]
    df_tasa = df_nacional[df_nacional['Tasas'] == tasa_actual]
    
    # Para que cada año sea una línea diferente, hacemos otro bucle for
    for anio in lista_anios:
        # Filtramos los datos de ese año y los ordenamos por trimestre (T1, T2...)
        df_anio = df_tasa[df_tasa['Año'] == anio].sort_values('Trimestre')
        
        # Dibujamos la línea de ese año con plt.plot
        plt.plot(df_anio['Trimestre'], df_anio['Total'], marker='o', label=anio)
    
    # Ponemos los textos de cada cuadro
    plt.title(tasa_actual)
    plt.xlabel('Trimestre')
    plt.ylabel('Tasa (%)')
    
    # Ponemos la leyenda (el cuadro con los años)
    # Usamos 'prop' para que la letra sea más pequeña y no tape el gráfico
    plt.legend(title="Año", fontsize='7', ncol=2)

# Ajustamos para mejorar la visualización y mostramos
plt.tight_layout()
plt.show()

```
